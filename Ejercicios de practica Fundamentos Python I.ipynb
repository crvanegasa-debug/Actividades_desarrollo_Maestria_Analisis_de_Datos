```python
# Del pseudocódigo a Python
# Conserva la lógica y cambia la representación
# Traduce una decisión
# Completa el código Python para que represente exactamente el pseudocódigo mostrado.
# PSEUDOCÓDIGO
# LEER nota
# SI nota >= 3.0 ENTONCES
# ESCRIBIR "Aprobado"
# SI NO
# ESCRIBIR "No aprobado"
#FIN SI
#PYTHON
nota = float(input("Nota: "))
if nota >= 3.0:
  print("Aprobado")
else:
  print("No aprobado")
```
